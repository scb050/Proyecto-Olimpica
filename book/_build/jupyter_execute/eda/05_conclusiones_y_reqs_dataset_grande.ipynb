## **5.1 Hallazgos clave del análisis exploratorio**

```python
from pathlib import Path
import pandas as pd

def find_project_root(start: Path, marker: str = "data", max_up: int = 6) -> Path:
    p = start.resolve()
    for _ in range(max_up):
        if (p / marker).exists():
            return p
        p = p.parent
    return start.resolve()

ROOT = find_project_root(Path.cwd())
DATA_PROCESSED = ROOT / "data" / "processed"

ts_path = DATA_PROCESSED / "ts_ventas_dia_tienda_producto.csv"

df_ts = pd.read_csv(ts_path, parse_dates=["FECHA"])

print("Shape:", df_ts.shape)
df_ts.head()


hallazgos = {
    "Total Ventas Netas": df_ts["VENTA_NETA"].sum(),
    "Total Unidades": df_ts["UNIDADES"].sum(),
    "Porcentaje promedio de descuento": (
        (df_ts["DESCUENTO"] / df_ts["VENTA_BRUTA"]).mean() * 100
        if "VENTA_BRUTA" in df_ts.columns else None
    ),
    "Productos distintos": df_ts["PLU_SAP"].nunique(),
    "Tiendas distintas": df_ts["CENTRO"].nunique(),
    "Fechas disponibles": df_ts["FECHA"].nunique()
}

hallazgos
```

Tras la auditoría y el análisis exploratorio realizado sobre la muestra de datos proporcionada por Olímpica, se identificaron los siguientes hallazgos relevantes:

- El dataset presenta una estructura transaccional a nivel de línea de factura (producto–ticket–tienda–fecha), lo cual es ideal para agregaciones temporales.
- Existe señal clara de actividad promocional mediante las variables de descuento y oferta.
- No se detectaron problemas estructurales graves (nulos masivos, inconsistencias críticas o valores imposibles).
- Se evidencia concentración de ventas en un subconjunto reducido de productos (patrón tipo 80/20).
- La variabilidad entre productos sugiere distintos comportamientos de demanda.

Estos resultados confirman que el dataset es técnicamente viable para la construcción de modelos predictivos.

## **5.2 Diagnóstico de calidad del dataset**

Desde el punto de vista técnico, la muestra analizada presenta:

###  Fortalezas
- Integridad estructural adecuada.
- Ausencia de duplicados críticos.
- Coherencia entre descuentos y promociones.
- Variables suficientes para análisis inicial de ventas.

###  Aspectos a considerar
- La muestra contiene únicamente una fecha, lo cual limita el análisis temporal.
- No se dispone de precio regular explícito.
- No se cuenta con información de inventario o disponibilidad.
- No hay segmentación geográfica ampliada en esta muestra.

En conclusión, la calidad es adecuada para Fase 1 (EDA), pero insuficiente aún para modelado temporal completo.

## **5.4 Limitaciones de la Muestra Actual**

La muestra proporcionada tiene las siguientes limitaciones:

- Solo una fecha disponible (no permite análisis de estacionalidad ni tendencia).
- No es posible calcular lags ni ventanas móviles reales.
- No se puede estimar elasticidad real sin histórico.
- No se puede evaluar efecto acumulado de promociones.

Por lo tanto, esta fase debe entenderse como una validación estructural preliminar.

## **5.5 Requerimientos Técnicos para el Dataset Completo**

Para avanzar hacia la Fase 2 (modelado predictivo de ventas), se recomienda solicitar:

###  Histórico
- Mínimo 2–3 años de información diaria.
- Granularidad Día – Tienda – Producto.

###  Variables adicionales clave
- Precio regular del producto.
- Tipo de promoción (porcentaje, 2x1, combo, etc.).
- Calendario comercial (festivos, campañas).
- Indicador de inventario o quiebres.
- (Opcional) Costo del producto.

###  Dimensiones complementarias
- Información geográfica de tiendas.
- Clasificación ampliada de productos.

Con estas variables, el modelo podrá capturar patrones reales de comportamiento comercial.
